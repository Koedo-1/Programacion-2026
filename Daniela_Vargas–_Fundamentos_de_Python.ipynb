# Problema #1: Calculadora de propina inteligente

## Problema

Dilan y sus amigos salieron a cenar y el restaurante les entregó una cuenta por ₡45000 .El grupo acordó dejar propina, pero cada persona sugirió un porcentaje diferente: 10%, 15% y 20%.

## Entrada
cuanto es el total de la cuenta, cuanta propina desean dejar y entre cuantos amigos se piensan dividir la cuenta

## Proceso
calcular la propina , el total final la cuenta y cuanto debe pagar cada uno


## Salida esperada

Va a mostrar el monto total de la cuenta incluyendo la propina y si la dividen entre amigos cuanto seria por cada uno

```python
#Entradas
try:
  total_cuenta = float(input("Ingrese el total de la cuenta: "))
except ValueError:
  print("el total de la cuenta debe ser un número mayor que 0.")
  total_cuenta = -1

try:
  porcentaje_quieren_dejar = float(input("Ingrese el porcentaje de propina que desean dejar: "))
except ValueError:
  print("el porcentaje de propina debe ser un número positivo.")
  porcentaje_quieren_dejar = -1

try:
  cantidad_amigos = int(input("Entre cuantos amigos van a divir la cuenta?: "))
except ValueError:
  print("¿cuantos amigos son?")
  cantidad_amigos = -1

#Proceso
if total_cuenta > 0 and porcentaje_quieren_dejar >=0 and cantidad_amigos > 0:
  propina = total_cuenta * porcentaje_quieren_dejar / 100
  total_final = total_cuenta + propina
  pago_persona = total_final / cantidad_amigos



#salidas
  print("-" * 50, "\n")
  print(f"Propina: ₡{propina:,.2f}")
  print(f"Total final: ₡{total_final:,.2f}")
  print(f"Cada persona debe pagar: ₡{pago_persona:,.2f}")
else:
  print("-" * 50, "\n")

  print("Hay datos inválidos. Revise lo que está poniendo.")
```

# Problema #2: Detector de número par/impar

## Problema

En una escuela están organizando una actividad para separar a los estudiantes en dos equipos. Para hacerlo, cada estudiante debe escribir un número entero en una tarjeta.

Las reglas son:

Si el número es par, el estudiante va al equipo azul.

Si el número es impar, el estudiante va al equipo rojo.

El programa debe pedir un número entero y decir a qué equipo pertenece.

## Entrada

- numero que el niño va a escribir

## Proceso

- Convertir lo escrito a número entero.

Verificar si es divisible entre 2.

Si el residuo es 0, es par.

Si el residuo no es 0, es impar.

## Salida esperada

- El programa deberia decir a que equipo va cada nino con el numero que ingresen

```python
#Entradas
try:
  numero_nino = int(input("Escriba el numero que le gustaria usar: "))
  if numero_nino % 2 == 0:
    print("-" * 50, "\n")
    print("Eres del equipo Azul!! ")
  else:
    print("-" * 50, "\n")
    print("Eres del equipo Rojo!!")

except ValueError:
  print("-" * 50, "\n")
  print("Escribe el número usando dígitos, por ejemplo: 3!")

#proceso
#% significa ¿cuánto sobra al dividir este número entre 2?
```

# Problema #4: Validador de compra con descuento

## Problema

Una tienda de maquillaje quiere calcular el total a pagar de sus clientas y aplicar descuentos según el monto de la compra y si son miembros del club de belleza.

## Entrada

- El programa debe solicitar:

Nombre de la clienta.

Monto de la compra.

Si es miembro del club de belleza (sí o no).

## Proceso

-Las reglas de descuento son:

Condición	Descuento

Compra de ₡30,000 o más y es miembro	20%

Compra de ₡30,000 o más y no es miembro	10%

Compra menor a ₡30,000 y es miembro	5%

Compra menor a ₡30,000 y no es miembro	0%


## Salida esperada

-El programa debe mostrar:

Nombre de la clienta.

Monto original de la compra.

Descuento aplicado en colones.

Total final a pagar.

Además, debe validar que el monto no sea negativo. Si se ingresa un valor inválido, debe mostrar un mensaje de error.

## Pruebas manuales


```python
# Entradas
nombre_clienta = input("Ingrese su nombre: ").capitalize().strip()

try:
    monto_compra = float(input("¿Cuál fue el total de su compra?: "))
except ValueError:
    print("Escriba el monto en dígitos, por ejemplo: 25000")
    monto_compra = -1

club = input("¿Pertenece al Club de Belleza? SI/NO: ").strip().lower() == "si"

# Procesos y salida
if monto_compra < 1:
    print("El monto de la compra debe ser mayor a ₡0.")
else:
    descuento = 0
    porcentaje_descuento = 0

    if monto_compra >= 30000 and club:
        porcentaje_descuento = 20
        descuento = monto_compra * 0.20
    elif monto_compra >= 30000 and not club:
        porcentaje_descuento = 10
        descuento = monto_compra * 0.10
    elif monto_compra < 30000 and club:
        porcentaje_descuento = 5
        descuento = monto_compra * 0.05
    else:
        porcentaje_descuento = 0
        descuento = 0

    total_pagar = monto_compra - descuento

    # Salida
    print(f"{nombre_clienta}, su compra original fue de ₡{monto_compra:,.2f}, su descuento es del {porcentaje_descuento}%, equivalente a ₡{descuento:,.2f}, y su monto final sería de ₡{total_pagar:,.2f}.")
```

¿Qué error encontró durante las pruebas?

¿Cómo mejoró la limpieza del código la calidad del programa?

Además, permitió separar los cálculos de los mensajes que se muestran al usuario, haciendo el programa más claro y fácil de modificar
